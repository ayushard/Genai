## 2. Load a Transformer model

```python
from transformers import pipeline

# Text generation model.
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

print("Model loaded successfully.")
```

## 3. Zero-shot style prompt

```python
prompt = "Artificial Intelligence is important because"

result = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=1,
    do_sample=False
)

print("Prompt:", prompt)
print("\nGenerated text:")
print(result[0]["generated_text"])
```

## 4. One-shot prompting

```python
one_shot_prompt = """
Example:
Question: What is Python?
Answer: Python is a programming language.

Question: What is Machine Learning?
Answer:
"""

result = generator(
    one_shot_prompt,
    max_new_tokens=35,
    num_return_sequences=1,
    do_sample=False
)

print(result[0]["generated_text"])
```

## 5. Few-shot prompting

```python
few_shot_prompt = """
Classify the sentiment.

Text: I love this product.
Sentiment: Positive

Text: This service is excellent.
Sentiment: Positive

Text: The product is terrible.
Sentiment: Negative

Text: The application is very useful.
Sentiment:
"""

result = generator(
    few_shot_prompt,
    max_new_tokens=20,
    num_return_sequences=1,
    do_sample=False
)

print(result[0]["generated_text"])
```

## 6. Text summarization using Hugging Face

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)

text = """
Generative Artificial Intelligence can create new content such as text,
images, audio and code. Transformer models are widely used in modern
generative AI because self-attention helps them understand relationships
between tokens. Large language models are trained on large text datasets
and can perform generation, summarization, question answering and other
natural language tasks.
"""

summary = summarizer(
    text,
    max_length=60,
    min_length=20,
    do_sample=False
)

print("Original text:\n", text)
print("\nSummary:\n", summary[0]["summary_text"])
```

## 7. Prompt engineering comparison

```python
prompts = {
    "Zero-shot": "Explain machine learning in two sentences.",
    "One-shot": """
Example: Explain AI in one sentence -> AI enables computers to perform tasks
that normally require human intelligence.
Now explain machine learning in one sentence ->
""",
    "Few-shot": """
Example 1: Python -> Programming Language
Example 2: Delhi -> City
Example 3: Java -> Programming Language
Example 4: Nagpur -> City
Example 5: TensorFlow ->
"""
}

for name, prompt in prompts.items():
    result = generator(
        prompt,
        max_new_tokens=25,
        num_return_sequences=1,
        do_sample=False
    )
    print("=" * 60)
    print(name)
    print(result[0]["generated_text"])
```

## Result
A Transformer-based Hugging Face model was used for text generation and summarization. Zero-shot, one-shot and few-shot prompting were demonstrated.
